## Darcy's Law — Basic Concept

The fundamental equation governing fluid flow in porous media:

$$q = -\frac{kA}{\mu} \frac{\Delta P}{L}$$

Where:
- $q$ = volumetric flow rate [m³/s]
- $k$ = permeability [m²] (or Darcy)
- $A$ = cross-sectional area [m²]
- $\mu$ = dynamic viscosity [Pa·s]
- $\Delta P$ = pressure difference [Pa]
- $L$ = length [m]

```python
def darcy_flow(k, A, mu, delta_p, L):
    """Calculate volumetric flow rate via Darcy's Law.
    
    Args:
        k: permeability in m² (1 Darcy = 9.869e-13 m²)
        A: cross-sectional area in m²
        mu: dynamic viscosity in Pa·s
        delta_p: pressure difference in Pa
        L: length in m
    Returns:
        q: volumetric flow rate in m³/s
    """
    return (k * A * delta_p) / (mu * L)

# Example: 100 mD sandstone, 1m² area, water, 1 bar over 100m
DARCY = 9.869e-13  # m²
k   = 100e-3 * DARCY  # 100 millidarcy
A   = 1.0             # m²
mu  = 1e-3            # Pa·s (water at 20°C)
dP  = 1e5             # Pa (1 bar)
L   = 100.0           # m

q = darcy_flow(k, A, mu, dP, L)
print(f'Darcy flow rate: {q:.3e} m³/s')
print(f'Darcy flow rate: {q*1e6:.4f} mL/s')
```
